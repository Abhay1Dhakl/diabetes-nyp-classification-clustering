## Import Models

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV file into a DataFrame
RAW_PATH = "../data/raw/diabetes.csv"
PROCESSED_PATH = "../data/processed/diabetes_clean.csv"

df = pd.read_csv(RAW_PATH)

# quick check of the data
print("shape of the data:", df.shape)
print("columns in the data:", df.columns)
print("data types of the columns:", df.dtypes)

display(df.head())
```

```python
# trim columns for better readability
df.columns = df.columns.str.strip()

# clean labels
df["CLASS"] = df["CLASS"].astype(str).str.strip()
df["Gender"] = df["Gender"].astype(str).str.strip().str.upper()

print("Class Values:", df["CLASS"].unique())
print("CLASS ounts:\n", df["CLASS"].value_counts())

print("Gender Values:\n", df["Gender"].unique())
print("Gender Counts:\n", df["Gender"].value_counts())
```

```python
print("Missing values per column:\n", df.isna().sum().sort_values(ascending=False))
display(df.describe(include="all"))
```

```python
id_cols = [c for c in ["ID", "No_Pation"] if c in df.columns]
X_df = df.drop(columns=id_cols, errors="ignore")
print("Dropped ID columns:", id_cols)
print("Remaining columns:", list(X_df.columns))
```

```python
df.to_csv(PROCESSED_PATH, index=False)
print("Saved:", PROCESSED_PATH)
```

# Step 2: EDA Analysis

## 1. Check for outliers

```python
number_columns = X_df.select_dtypes(include="number").columns

display(X_df[number_columns].describe())

print("Negative BMI values:", (X_df["BMI"] < 0).sum() if "BMI" in X_df else "BMI Missing")
print("Age <=0:", (X_df["AGE"] <= 0).sum() if "AGE" in X_df else "AGE Missing")
```

## 2. Univariate plots

```python
cols = [c for c in ["HbA1c","BMI","AGE","Chol","TG","LDL","HDL","Cr","Urea"] if c in X_df.columns]

X_df[cols].hist(bins=20, figsize=(12, 8))
plt.tight_layout()
plt.show()
```

```python
cols = [c for c in ["HbA1c","BMI","AGE","Chol","TG","LDL","HDL"] if c in X_df.columns]

for c in cols:
    plt.figure(figsize=(6,4))
    for label in sorted(df["CLASS"].unique()):
        df[df["CLASS"]==label][c].plot(kind="kde", label=label)
    plt.title(f"{c} distribution by CLASS")
    plt.legend()
    plt.show()
```

```python
import numpy as np

corr = X_df[number_columns].corr(numeric_only=True)
plt.figure(figsize=(10,7))
plt.imshow(corr, aspect="auto")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar()
plt.title("Correlation heatmap (numeric features)")
plt.tight_layout()
plt.show()
```
